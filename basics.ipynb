## Conditionals

```python
# multiple if statements
number = 10
if number > 5:
    print("Number is greater than 5")
if number >8:
    print("Number is greater than 8")

print()

# if-elif statements
if number > 5:
    print("Number is greater than 5")
elif number >8:
    print("Number is greater than 8")
```

```python
import random
number = random.randint(1,5)
tries = 3
while True:
    guess = int(input("What is the number? "))
    if guess == number:
        print("You guessed the number!")
        break
    else:
        tries-=1
        print("Try again!")
        print("You gave ", tries, "tries left")
        if tries == 0:
            print("You have used all your tries!")
            break
```

## Exceptions

```python
# try, except, else, finally
while True:
    try:
        number = int(input("What is number? "))
    except ValueError:
        print("Not a number")
    else:
        print(f"{number} is a number")
        break
    finally:
        print("This will always be executed")
```
