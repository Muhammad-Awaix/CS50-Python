```python
import random
number = random.randint(1,5)
tries = 3
while True:
    guess = int(input("What is the number? "))
    if guess == number:
        print("You guessed the number!")
        break
    else:
        tries-=1
        print("Try again!")
        print("You gave ", tries, "tries left")
        if tries == 0:
            print("You have used all your tries!")
            break
```

```python
score = int(input("What is your score? "))
if 0 < score > 100:
    print("Invalid score")
elif 90 <= score:
    print("Grade: A")
elif 80 <= score:
    print("Grade: B")
elif 60 <= score:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
def main():
    x = int(input("What is x? "))
    if is_even(x):
        print("x is even")
    else:
        print("x is odd")
 
def is_even(n):
    return n % 2 ==0

main()
```
