```python
import random
number = random.randint(1,5)
tries = 3
while True:
    guess = int(input("What is the number? "))
    if guess == number:
        print("You guessed the number!")
        break
    else:
        tries-=1
        print("Try again!")
        print("You gave ", tries, "tries left")
        if tries == 0:
            print("You have used all your tries!")
            break
```

## Exceptions

```python
# try, except, else, finally
while True:
    try:
        number = int(input("What is number? "))
    except ValueError:
        print("Not a number")
    else:
        print(f"{number} is a number")
        break
    finally:
        print("This will always be executed")
```

```python
krish_web = 'https://www.krishnaik.in/'
k_content = requests.get(krish_web).content
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(k_content, 'html.parser')
print(soup.prettify())
```

```python
print(soup.find_all('img'))
```

**The 80/20 Rule of Beautiful Soup**
To be a pro, you only need to master the 20% of features that you will use 80% of the time:
- *BeautifulSoup(markup, 'html.parser'): Always start here to load your data.*
- *find() and find_all(): Use these to grab specific tags or lists of tags by name, class, or ID.*
- *Attribute Access (tag['href']): This is how you get links out of **a** tags or image sources out **img** of tags.*
- *get_text(): The quickest way to get the actual "story" or information out of the website without the messy code.*
- *prettify(): Use this while you are writing your code to see what the "toy box" looks like so you don't get lost.*
